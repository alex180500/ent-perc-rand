```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.load("big_simulations/uniform_grid.npz")
windows, prob, lcc = data["windows"][:-1], data["prob"][:-1], data["lcc"][:-1]
```

```python
# %matplotlib qt
plt.rcParams.update({'font.size': 15})
plt.style.use("bmh")
plt.figure(figsize=(8, 6), dpi=200, layout="constrained", facecolor="None")
# plt.axvline(0.5, c="k", zorder=50, lw=2)
n = len(windows)
colors = plt.cm.jet(np.linspace(0, 1, n))
for idx in range(n):
    plt.plot(prob[idx], lcc[idx], label=f"{windows[idx]*2:.1f}", c=colors[idx], lw=4)
plt.legend(title="Width")
# plt.title("Uniform Probability within a window Percolation")
plt.axvline(0.5, c="k", lw=3)
plt.xlim(0, 1)
plt.ylabel(r"Percolation Strength $P_\infty$")
plt.xlabel(r"Average Probability $\langle p \rangle$")
plt.savefig("../thesis graphics/graphics/random-grid-simulation.pdf", bbox_inches="tight")
```
